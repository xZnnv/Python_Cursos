Por ejemplo:

- escribir su nombre

## ❓ Pregunta

Si queremos que el usuario escriba información sobre el celular…

¿Cómo hacemos para que el programa (código) la reciba?

## 💡 Concepto

**INPUT**

`input()` (entrada) permite que el programa reciba información del usuario.

---

### Pista:
Si con `print()` el celular habla…

👉 con `input()` el celular escucha.

## Código Base

```python
# Vamos a introducir un dato muy importante!
# Quien es el dueño (owner) de este celular.
owner = input("Who is the owner of the cellphone? ")

# Solo para verficar que el código leyo/escucho bien
# el dato que queremos introducir, li imprimimos (print)
# en pantalla.
print(owner)
```

## Código Expandido

```python
# Continuamos introduciendo información de nuestro Cell Phone (Celular)
brand = input("What is the brand of the cellphone? ")
color = input("What color is the cellphone? ")

# Imprimimos (print) de manera más organizada la información.
print("Brand:", brand)
print("Color:", color)
```

## Mini Reto

```python
# Mini reto:
# crea dos nuevas preguntas sobre tu CellPhone
# puedes utilizar IA para generar las preguntas en
# ingles para mantener el estilo del Mini Reto y
# practiques tu entendimiento de inglés.
# Utiliza nombre de variables en inglés.

storage = input("¿How much storage capacity does your cell phone have? ")
processor = input("¿ what processor does your phone have ?")



battery = input("What is the battery level? ")
model = input("What is the model of the cellphone? ")
year = input("What is the year of the cellphone? ")


# Muestra la nueva información en la manera organizada
# como lo venimos haciendo.
print("Battery:", battery)
print("Model:", model)
print("Year:", year)
print("storage", storage)
print("processor", processor)
```

# 🔵 BLOQUE 4 — IF (decisión · lógica · comportamiento)

## 🧠 Activación

Nuestro **CellPhone** ya puede:

- guardar información (variable)
- mostrar información (print)
- recibir información (input)

Pero aún le falta algo importante…

👉 tomar decisiones.

Por ejemplo:

- si la batería está baja → mostrar alerta  
- si la batería está alta → todo está bien  

---

### Micro reto:
Piensa en tu celular…

👉 ¿qué decisiones toma automáticamente todos los días?

![If_CellPhone](https://drive.google.com/uc?export=view&id=1NZNIL6lSn1EBm2nII_P33EdIue4hByIS)

## ❓ Pregunta

Si el celular tiene información…

¿Cómo hacemos para que el programa (código) tome decisiones con ella?

## 💡 Concepto

**IF**

`if` (si) permite que el programa tome decisiones cuando una condición se cumple.

---

### Pista:
Si con `input()` el celular escucha…

👉 con `if` el celular decide qué hacer.

## Código Base

```python
# Nivel de batería de nuestro Cell Phone
# Puedes variar el valor de la variable 'battery'
# para probar si la alerta se imprime (print) o no.
battery = 15

# Si (if) la batería es menor a 20
if battery < 20:
    # Imprimimos (print) una alerta
    print("Low battery")
```

## Código Expandido

```python
# Continuamos trabajando con nuestro Cell Phone
# Puedes variar el valor de la variable 'battery'
# para probar si el mensaje se imprime (print) o no.
battery = 85

# Tomamos una decisión según el nivel de batería
if battery > 20:
    print("Phone is ready")
```

## Mini Reto

```python
# Mini reto:
# crea una condición para tu Cell Phone

battery = 0

# Define un valor real para battery
# y crea una condición usando if

if battery < 20:
    print("Low battery")

# Crea una segunda condición
# con otro mensaje, el mensaje debe
# estar en inglés, puedes ayudarte con IA.

wifi = input("Are you connected to a Wi-Fi network:  ")

if wifi == "yes":
  print("You can browse the networks")
```

## 🧩 Cierre

Con `if` (si), el programa (código) deja de solo guardar, mostrar o recibir información.

Ahora puede tomar decisiones.

Nuestro **CellPhone ya puede evaluar información (de forma básica)** para decidir qué hacer.
